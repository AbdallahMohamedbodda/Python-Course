**4- Make a "Generic" Code that by taking any Sentence as an input it will return two Sentences which are splitted by the Middle Letter.**

Examples :- input = "harpy" ,output= ["ha" ,"py"]

input="Mastery" ,output=["Mas","ery"]

input="He was Lucky" ,output=['He wa', ' Lucky']

```python
def split_by_middle(sentence):
    # Remove leading/trailing whitespace
    sentence = sentence.strip()
    
    # Handle empty string case
    if not sentence:
        return ("", "")
    
    length = len(sentence)
    middle_index = length // 2
    
    # For even length, split at middle
    # For odd length, include the middle character in both parts
    if length % 2 == 0:
        part1 = sentence[:middle_index]
        part2 = sentence[middle_index:]
    else:
        part1 = sentence[:middle_index + 1]
        part2 = sentence[middle_index:]
    
    return (part1, part2)

# Example usage
input_sentence = input("Enter a sentence: ")
part1, part2 = split_by_middle(input_sentence)

print("First part:", part1)
print("Second part:", part2)
```

**6-Write a Python program which accepts a sequence of comma-separated numbers from user and generate a list and a tuple with those numbers.**

```python
# Accept comma-separated numbers from user
numbers_input = input("Enter comma-separated numbers: ")

# Split the input string into individual number strings
number_list = numbers_input.split(',')

# Convert each string to a number (float to handle decimals)
numbers = [float(num.strip()) for num in number_list]

# Generate the tuple
numbers_tuple = tuple(numbers)

# Display the results
print("List:", numbers)
print("Tuple:", numbers_tuple)
```
