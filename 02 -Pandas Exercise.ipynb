**Import pandas and read in the 'data/Ecommerce Purchases.csv' file and set it to a DataFrame called ecom.**

```python
import pandas as pd

ecom = pd.read_csv('data/Ecommerce Purchases.csv')

```

**Check the head of the DataFrame.**

```python
print(ecom.head())  # Displays the first 5 rows of the DataFrame
```

**How many rows and columns are there?**

```python

rows, columns = ecom.shape
print(f"There are {rows} rows and {columns} columns in the DataFrame.")

```

**What is the average Purchase Price?**

```python
average_price = ecom['Purchase Price'].mean()
print(f"The average Purchase Price is: {average_price}")
```

**What were the highest and lowest purchase prices?**

```python
highest_price = ecom['Purchase Price'].max()


print(f"Highest Purchase Price: {highest_price}")
print(f"Lowest Purchase Price: {lowest_price}")
```

```python
lowest_price = ecom['Purchase Price'].min()
lowest_price = ecom['Purchase Price'].min()
```

**How many people have English 'en' as their Language of choice on the website?**

```python
english_users = ecom[ecom['Language'] == 'en'].shape[0]
print(f"Number of people with English as their language: {english_users}")
```

**How many people have the job title of "Lawyer" ?**

```python
lawyers_count = ecom[ecom['Job'] == 'Lawyer'].shape[0]
print(f"Number of people with the job title 'Lawyer': {lawyers_count}")
```

**How many people made the purchase during the AM and how many people made the purchase during PM ?**

**(Hint: Check out [value_counts()](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.value_counts.html) )**

```python
# Ensure 'Purchase Time' is in datetime format
ecom['Purchase Time'] = pd.to_datetime(ecom['Purchase Time'])

# Classify the purchases as AM or PM based on the hour
ecom['AM_PM'] = ecom['Purchase Time'].dt.hour.apply(lambda x: 'AM' if x < 12 else 'PM')

# Count how many people made purchases during AM and PM
purchase_times = ecom['AM_PM'].value_counts()

print(purchase_times)

```

**What are the 5 most common Job Titles?**

```python
# Get the top 5 most common job titles
top_5_job_titles = ecom['Job'].value_counts().head(5)

# Display the result
print("Top 5 most common job titles:")
print(top_5_job_titles)
```

**Someone made a purchase that came from Lot: "90 WT" , what was the Purchase Price for this transaction?**

```python
purchase_price_90wt = ecom[ecom['Lot'] == '90 WT']['Purchase Price'].values[0]
print(f"Purchase Price for Lot '90 WT': {purchase_price_90wt}")
```

**What is the email of the person with the following Credit Card Number: 4926535242672853**

```python
email = ecom[ecom['Credit Card'] == 4926535242672853]['Email'].values[0]
print(f"Email for credit card 4926535242672853: {email}")
```

**How many people have American Express as their Credit Card Provider *and* made a purchase above $95 ?**

```python
amex_high_purchases = ecom[(ecom['CC Provider'] == 'American Express') & (ecom['Purchase Price'] > 95)].shape[0]
print(f"American Express users with purchases over $95: {amex_high_purchases}")
```

**Hard: How many people have a credit card that expires in 2025?**

```python
cc_2025 = ecom[ecom['CC Exp Date'].apply(lambda x: x.split('/')[1] == '25')].shape[0]
print(f"Number of credit cards expiring in 2025: {cc_2025}")
```

**Hard: What are the top 5 most popular email providers/hosts (e.g. gmail.com, yahoo.com, etc...)**

```python
top_email_providers = ecom['Email'].apply(lambda x: x.split('@')[1]).value_counts().head(5)
print("Top 5 email providers:")
print(top_email_providers)
```
